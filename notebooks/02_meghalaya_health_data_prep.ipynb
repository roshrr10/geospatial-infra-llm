```python
# STEP 2.9: Load School Point Shapefile

schools_gdf = gpd.read_file("../data/raw/meghalaya/boundaries/school/schools.shp")
schools_gdf.head()
```

```python
# STEP 2.9.1: Standardize UDISE, Block, District column names

schools_gdf = schools_gdf.rename(columns={
    "udiseCode": "udise_code",
    "Block_Name": "block_name",
    "District_N": "district_name"
})

schools_gdf["udise_code"] = schools_gdf["udise_code"].astype(str)

schools_gdf.head()
```

```python
def clean_udise(x):
    try:
        return str(int(float(x)))
    except:
        return None

school_health["udise_code"] = school_health["udise_code"].apply(clean_udise)
schools_gdf["udise_code"] = schools_gdf["udise_code"].apply(clean_udise)

# Check
schools_gdf["udise_code"].head(), school_health["udise_code"].head()
```

```python
# STEP 2.10: Merge health indicators with school spatial layer

school_health_gdf = schools_gdf.merge(
    school_health,
    on="udise_code",
    how="left"
)

# Check result
print(school_health_gdf.shape)
school_health_gdf.head()
```

```python
# Check available columns related to IFA and Deworming
cols = school_health_gdf.columns
[c for c in cols if "ifa" in c.lower() or "girl" in c.lower() or "boy" in c.lower() or "treated" in c.lower()]
```

```python
# Convert required columns to numeric
num_cols = [
    "total_girls", "total_boys",
    "consuming_girls", "consuming_boys",
    "enrolled_girls", "enrolled_boys",
    "treated_girls", "treated_boys"
]

for col in num_cols:
    school_health_gdf[col] = pd.to_numeric(school_health_gdf[col], errors="coerce")
```

```python
# STEP 2.11: Compute coverage indicators

school_health_gdf["ifa_coverage_girls"] = (
    school_health_gdf["consuming_girls"] / school_health_gdf["total_girls"]
)

school_health_gdf["ifa_coverage_boys"] = (
    school_health_gdf["consuming_boys"] / school_health_gdf["total_boys"]
)

school_health_gdf["deworm_coverage_girls"] = (
    school_health_gdf["treated_girls"] / school_health_gdf["enrolled_girls"]
)

school_health_gdf["deworm_coverage_boys"] = (
    school_health_gdf["treated_boys"] / school_health_gdf["enrolled_boys"]
)

school_health_gdf[
    ["ifa_coverage_girls", "ifa_coverage_boys",
     "deworm_coverage_girls", "deworm_coverage_boys"]
].head()
```

```python
# STEP 2.12: Block-wise health aggregation

block_health = school_health_gdf.groupby("block_name_x").agg({
    "ifa_coverage_girls": "mean",
    "ifa_coverage_boys": "mean",
    "deworm_coverage_girls": "mean",
    "deworm_coverage_boys": "mean"
}).reset_index()

block_health.columns = [
    "block_name",
    "avg_ifa_coverage_girls",
    "avg_ifa_coverage_boys",
    "avg_deworm_coverage_girls",
    "avg_deworm_coverage_boys"
]

block_health.head()
```

```python
# STEP 3.1: Create Meghalaya state boundary from district shapefile

import geopandas as gpd

districts_shape = gpd.read_file("../data/raw/meghalaya/boundaries/district/district.shp")

# Dissolve all districts into one state polygon
state_shape = districts_shape.dissolve()

# Check result
state_shape.head(), state_shape.crs
```

```python
# STEP 3.2: Download Meghalaya road network from OpenStreetMap

import osmnx as ox

# Use the state boundary polygon
meghalaya_polygon = state_shape.geometry.iloc[0]

# Download all drivable roads
G = ox.graph_from_polygon(meghalaya_polygon, network_type="drive")

# Convert to GeoDataFrame (edges = road segments)
roads_gdf = ox.graph_to_gdfs(G, nodes=False)

# Inspect
roads_gdf.head(), roads_gdf.crs
```

```python
# STEP 3.3: Convert road network graph to GeoDataFrame
edges = ox.graph_to_gdfs(G, nodes=False, edges=True)
edges = edges.reset_index()

# Keep only required columns
roads_gdf = edges[["geometry", "highway", "length"]].copy()
roads_gdf.crs = "EPSG:4326"

roads_gdf.head(), roads_gdf.shape
```

```python
# STEP 3.4: Reproject both layers to metric CRS for accurate length (meters)
blocks_proj = blocks_shape.to_crs(epsg=32646)   # UTM zone for Meghalaya
roads_proj = roads_gdf.to_crs(epsg=32646)

# Spatial join: assign each road segment to a block
roads_with_blocks = gpd.sjoin(roads_proj, blocks_proj, how="left", predicate="intersects")

roads_with_blocks.head(), roads_with_blocks.shape
```

```python
# STEP 3.5: Drop roads not matched to any block
roads_with_blocks_clean = roads_with_blocks.dropna(subset=["block_name"])

# Convert length from meters to kilometers
roads_with_blocks_clean["length_km"] = roads_with_blocks_clean.geometry.length / 1000

# Aggregate total road length per block
block_road_stats = roads_with_blocks_clean.groupby("block_name").agg(
    total_road_km=("length_km", "sum"),
    block_area_sqkm=("area_sqkm", "first")
).reset_index()

# Compute road density (km of road per sq km of area)
block_road_stats["road_density_km_per_sqkm"] = (
    block_road_stats["total_road_km"] / block_road_stats["block_area_sqkm"]
)

block_road_stats.head()
```

```python
# STEP 3.6: Join road stats with block geometries

block_infra_gdf = blocks_shape.merge(
    block_road_stats,
    on="block_name",
    how="left"
)

block_infra_gdf.head()
```

```python
# Standardize block names to uppercase in both tables
schools_gdf["block_name"] = schools_gdf["block_name"].str.upper()
blocks_shape["block_name"] = blocks_shape["block_name"].str.upper()

# Recompute school counts
school_counts = schools_gdf.groupby("block_name").size().reset_index(name="total_schools")

# Merge with block areas
school_density = school_counts.merge(
    blocks_shape[["block_name", "area_sqkm"]],
    on="block_name",
    how="left"
)

# Compute density
school_density["schools_per_sqkm"] = school_density["total_schools"] / school_density["area_sqkm"]

school_density.head()
```

```python
road_density = block_road_stats[[
    "block_name",
    "total_road_km",
    "block_area_sqkm",
    "road_density_km_per_sqkm"
]]

road_density.head()
```

```python
road_density["block_name"] = road_density["block_name"].str.upper()
school_density["block_name"] = school_density["block_name"].str.upper()

infra_intelligence = road_density.merge(
    school_density,
    on="block_name",
    how="left"
)

infra_intelligence.head()
```

```python
# Ensure same case for joining
blocks_shape["block_name"] = blocks_shape["block_name"].str.upper()

block_infra_gdf = blocks_shape.merge(
    infra_intelligence,
    on="block_name",
    how="left"
)

block_infra_gdf.head()
```

```python
query = """
WITH thresholds AS (
    SELECT
        percentile_cont(0.25) WITHIN GROUP (ORDER BY road_density_km_per_sqkm) AS low_road_threshold,
        percentile_cont(0.75) WITHIN GROUP (ORDER BY schools_per_sqkm) AS high_schools_threshold
    FROM meghalaya_block_intelligence
),
filtered_blocks AS (
    SELECT *
    FROM meghalaya_block_intelligence
    WHERE road_density_km_per_sqkm < (SELECT low_road_threshold FROM thresholds)
      AND schools_per_sqkm >= (SELECT high_schools_threshold FROM thresholds)
)
SELECT block_name, district_name, road_density_km_per_sqkm, schools_per_sqkm
FROM filtered_blocks
ORDER BY road_density_km_per_sqkm ASC;
"""

result_df = pd.read_sql(query, conn)
result_df
```

```python
import pandas as pd
import geopandas as gpd

# Load final block intelligence
gdf = gpd.read_file("../data/processed/meghalaya/infra_health_master.gpkg", layer="block_intelligence")

# Normalize metrics (0–1 scale)
gdf["road_score"] = 1 - (gdf["road_density_km_per_sqkm"] / gdf["road_density_km_per_sqkm"].max())
gdf["school_score"] = gdf["schools_per_sqkm"] / gdf["schools_per_sqkm"].max()

# Simple priority index (can be weighted later)
gdf["infra_priority_score"] = (
    0.6 * gdf["road_score"] +
    0.4 * gdf["school_score"]
)

gdf[["block_name", "infra_priority_score"]].sort_values("infra_priority_score", ascending=False).head()
```

```python
import geopandas as gpd

gdf = gpd.read_file("../data/processed/meghalaya/infra_health_master.gpkg", layer="block_intelligence")
print(gdf.columns)
```

```python
# Fill missing health with 0 (means 0% coverage -> high priority)
health_cols = [
    "avg_ifa_coverage_girls",
    "avg_ifa_coverage_boys",
    "avg_deworm_coverage_girls",
    "avg_deworm_coverage_boys"
]

gdf[health_cols] = gdf[health_cols].fillna(0)

# Recompute components
gdf["road_priority"] = 1 - (gdf["road_density_km_per_sqkm"] / gdf["road_density_km_per_sqkm"].max())
gdf["school_pressure"] = gdf["schools_per_sqkm"] / gdf["schools_per_sqkm"].max()
gdf["health_gap"] = 1 - gdf[health_cols].mean(axis=1)

# Final composite priority score
gdf["infra_priority_score"] = (
    0.4 * gdf["road_priority"] +
    0.3 * gdf["school_pressure"] +
    0.3 * gdf["health_gap"]
)

# View top priority blocks
gdf[["block_name","infra_priority_score"]].sort_values(
    "infra_priority_score", ascending=False
).head()
```

```python
priority_path = "../data/processed/meghalaya/block_priority_intelligence.gpkg"

gdf.to_file(
    priority_path,
    layer="priority_blocks",
    driver="GPKG"
)

print("Saved at:", priority_path)
```
