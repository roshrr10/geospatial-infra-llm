```python
# Load district boundary
district_gdf = gpd.read_file("../data/raw/boundaries/district/HP_District_Boundary.shp")

# Ensure CRS is EPSG:4326
district_gdf = district_gdf.to_crs(epsg=4326)

district_gdf.head()
```

```python
district_gdf.columns
```

```python
# Rename geometry column
district_gdf = district_gdf.rename_geometry("geom")

# Select and rename useful columns
district_gdf = district_gdf[[
    "name",         # District name
    "To_Schools",   # Total schools
    "geom"
]].rename(columns={
    "name": "district_name",
    "To_Schools": "total_schools"
})

district_gdf.head()
```

```python
district_gdf.isnull().sum()
district_gdf.dtypes
```

```python
# Convert total_schools to numeric
district_gdf["total_schools"] = (
    district_gdf["total_schools"]
    .astype(str)                # ensure string
    .str.strip()                # remove spaces
    .replace("", "0")            # handle empty values
    .astype(int)                # convert to int
)

district_gdf.dtypes
```

```python
district_gdf.to_postgis(
    name="hp_district_boundary",
    con=engine,
    if_exists="replace",
    index=False
)
```

```python
import osmnx as ox
import networkx as nx

ox.__version__
```

```python
import geopandas as gpd

state_gdf = gpd.read_postgis(
    "SELECT state_name, geom FROM hp_state_boundary",
    con=engine,
    geom_col="geom"
)

state_gdf
```

```python
state_polygon = state_gdf.geometry.iloc[0]
```

```python
ox.settings.log_console = True
ox.settings.use_cache = True

G = ox.graph_from_polygon(
    state_polygon,
    network_type="drive"   # drivable roads
)
```

```python
type(G)
```

```python
len(G.nodes), len(G.edges)
```

```python
nodes, edges = ox.graph_to_gdfs(G)

edges = edges[["highway", "length", "geometry"]].copy()
edges = edges.rename(columns={"geometry": "geom"})

edges.head()
```

```python
edges["highway"].value_counts().head()
```

```python
edges = edges.set_geometry("geom")
```

```python
edges.to_postgis(
    name="hp_roads_raw",
    con=engine,
    if_exists="replace",
    index=False
)
```
