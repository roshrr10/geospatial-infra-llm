```python
import geopandas as gpd
from sqlalchemy import create_engine
```

```python
district_path = "../data/raw/meghalaya/boundaries/district/district.shp"  # change name if needed

gdf = gpd.read_file(district_path)
gdf.head()
```

```python
print(gdf.crs)
```

```python
gdf = gdf.to_crs(epsg=4326)
print(gdf.crs)
```

```python
gdf.columns
```

```python
gdf["district_name"] = gdf["name"].astype(str).str.strip().str.title()
gdf[["id", "name", "district_name"]].head()
```

```python
from sqlalchemy import create_engine

engine = create_engine("postgresql://postgres:7654@localhost:5433/meghalaya_geo_llm")

engine
```

```python
gdf.to_postgis(
    name="meghalaya_districts",
    con=engine,
    if_exists="replace",
    index=False
)
```
