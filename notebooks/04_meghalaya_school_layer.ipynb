```python
import geopandas as gpd
from sqlalchemy import create_engine
```

```python
school_path = "../data/raw/meghalaya/boundaries/school/schools.shp"

schools_gdf = gpd.read_file(school_path)

schools_gdf.head()
```

```python
print(schools_gdf.crs)
```

```python
from sqlalchemy import create_engine

engine = create_engine("postgresql://postgres:7654@localhost:5433/meghalaya_geo_llm")

engine
```

```python
schools_gdf.to_postgis(
    name="meghalaya_schools",
    con=engine,
    if_exists="replace",
    index=False
)
```
